```python
from sklearn.preprocessing import OneHotEncoder
```

```python
# One Hot encoding
Airline_onehot_encoder_obj = OneHotEncoder(sparse=False)
Airline_onehot_encoder = Airline_onehot_encoder_obj.fit_transform(test_df['Airline'].to_numpy().reshape(-1, 1))

Source_onehot_encoder_obj = OneHotEncoder(sparse=False)
Source_onehot_encoder = Source_onehot_encoder_obj.fit_transform(test_df['Source'].to_numpy().reshape(-1, 1))

Destination_onehot_encoder_obj = OneHotEncoder(sparse=False)
Destination_onehot_encoder = Destination_onehot_encoder_obj.fit_transform(test_df['Destination'].to_numpy().reshape(-1, 1))

Additional_Info_onehot_encoder_obj = OneHotEncoder(sparse=False)
Additional_Info_onehot_encoder = Additional_Info_onehot_encoder_obj.fit_transform(test_df['Additional_Info'].to_numpy().reshape(-1, 1))
```

```python
# dataframes for each encoding

airline_df = pd.DataFrame(
    Airline_onehot_encoder, # encoder
    columns = Airline_onehot_encoder_obj.get_feature_names(['Airline'])
)

source_df = pd.DataFrame(
    Source_onehot_encoder,
    columns = Source_onehot_encoder_obj.get_feature_names(['Source'])
)

destination_df = pd.DataFrame(
    Destination_onehot_encoder,
    columns = Destination_onehot_encoder_obj.get_feature_names(['Destination'])
)

additional_info_df = pd.DataFrame(
    Additional_Info_onehot_encoder,
    columns = Additional_Info_onehot_encoder_obj.get_feature_names(['Additional_Info'])
)
```
